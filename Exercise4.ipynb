## A. Time-Based Visualizations

1. Global Trend `(5 pts)`

Aggregate the data by Date and create a line chart showing the global number of confirmed COVID-19 cases over time.

```python
# put your answer here
import pandas as pd
import matplotlib.pyplot as plt

df["Date"] = pd.to_datetime(df["Date"])

global_confirmed = (
    df.groupby("Date")["Confirmed"]
    .sum()
    .reset_index()
    .sort_values("Date")
    )

plt.figure()
plt.plot(global_confirmed["Date"], global_confirmed["Confirmed"])
plt.xlabel("Date")
plt.ylabel("Confirmed Cases")
plt.title("Global Confirmed COVID-19 Cases Over Time")
plt.show()
```

2. Country-Level Trends `(5 pts)`

Select three countries and visualize their confirmed case counts over time on the same plot.

```python
# put your answer here
import matplotlib.pyplot as plt

df["Date"] = pd.to_datetime(df["Date"])

countries = ["Philippines", "US", "India"]

country_df = df[df["Country/Region"].isin(countries)]

country_trends = (
    country_df.groupby(["Date", "Country/Region"])["Confirmed"]
    .sum()
    .reset_index()
    )

# Plot
plt.figure()

data = country_trends[country_trends["Country/Region"] == "Philippines"]
plt.plot(data["Date"], data["Confirmed"], label="Philippines")
data = country_trends[country_trends["Country/Region"] == "US"]
plt.plot(data["Date"], data["Confirmed"], label="US")
data = country_trends[country_trends["Country/Region"] == "India"]
plt.plot(data["Date"], data["Confirmed"], label="India")

plt.xlabel("Date")
plt.ylabel("Confirmed Cases")
plt.title("Confirmed COVID-19 Cases Over Time (Selected Countries)")
plt.legend()
plt.show()
```

3. Active vs Recovered `(5 pts)`

For a selected country, create a line chart showing Active and Recovered cases over time.

```python
# put your answer here
selected_country = "Philippines"

country_data = df[df["Country/Region"] == selected_country]

country_daily = (
    country_data.groupby("Date")[["Active", "Recovered"]]
    .sum()
    .reset_index()
    )
# Plot
plt.figure()
plt.plot(country_daily["Date"], country_daily["Active"], label="Active Cases")
plt.plot(country_daily["Date"], country_daily["Recovered"], label="Recovered Cases")

plt.xlabel("Date")
plt.ylabel("Number of Cases")
plt.title(f"Active vs Recovered COVID-19 Cases in {selected_country}")
plt.legend()
plt.show()
```

## B: Comparative Visualizations

4. Country Comparison `(5 pts)`

Using data from a single date, create a bar chart showing the top 10 countries by confirmed cases.

```python
# put your answer here
latest_date = df["Date"].max()

latest_data = df[df["Date"] == latest_date]

top_countries = (
    latest_data.groupby("Country/Region")["Confirmed"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    )
# Plot
plt.figure()
top_countries.plot(kind="bar")

plt.xlabel("Country")
plt.ylabel("Confirmed Cases")
plt.title(f"Top 10 Countries by Confirmed COVID-19 Cases ({latest_date.date()})")
plt.show()
```

Aggregate confirmed cases by WHO Region and visualize the result using a bar chart.

## C. Geographic Visualization

6. Geographic Spread `(10 pts)`

Using Latitude and Longitude, create a map-based visualization showing confirmed cases for a selected date.

```python
# put your answer here
import plotly.express as px
map_date = df[df["Date"] == "2020-05-11"]

fig = px.choropleth(
    map_date,
    locations="Country/Region",
    locationmode="country names",
    color="Confirmed",
    hover_name="Country/Region",
    title="Geographic Spread of COVID-19 Cases (2020-05-11)"
)

fig.show()
```

7. Regional Clustering `(15 pts)`

Create a visualization that shows how confirmed cases are distributed geographically within a single WHO Region.

```python
# put your answer here
df["Confirmed Cases per Region"] = df.groupby("WHO Region")["Confirmed"].transform("sum")
region_cases = df[['WHO Region', 'Country/Region', 'Confirmed Cases per Region']]
region_cases
```

```python
fig = px.choropleth(
    region_cases,
    locations="Country/Region",
    locationmode="country names",
    color="Confirmed Cases per Region",
    hover_name="Country/Region",
    animation_frame="WHO Region",
    title="Geographic Confirmed Cases of COVID-19 per WHO Region"
)

fig.update_layout(margin={"r":0, "t":40, "l":0, "b":0})
fig.show()
```
